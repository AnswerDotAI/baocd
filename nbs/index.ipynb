# baocd

> BasedPL Python port of [advent-of-code-data](https://github.com/wimglenn/advent-of-code-data) for SolveIt and Jupyter notebooks.

`baocd` is a BasedPL Python (`basedpl`) port of [advent-of-code-data](https://github.com/wimglenn/advent-of-code-data) for loading Advent of Code puzzle data and submitting answers in SolveIt and Jupyter notebooks.

## Setup

1. Login to [Advent of Code](https://adventofcode.com)
2. Open browser Dev Tools and copy the `session` cookie.
3. Set it as `AOC_SESSION` in your SolveIt secrets.
