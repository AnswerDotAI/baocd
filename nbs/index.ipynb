# baocd

> BasedPL Python port of [advent-of-code-data](https://github.com/wimglenn/advent-of-code-data) for SolveIt.

`baocd` is a [BasedPL](https://github.com/AnswerDotAI/basedpl) port of [advent-of-code-data](https://github.com/wimglenn/advent-of-code-data) for AoC puzzle data in SolveIt.



## Installation

```bash
 pip install git+https://github.com/AnswerDotAI/baocd.git
```


## Setup


1. Log in to [Advent of Code](https://adventofcode.com)
2. Open browser Dev Tools and copy the `session` cookie.
3. Set it as `AOC_SESSION` in your SolveIt secrets.
